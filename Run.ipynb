```python
# Write debug inference script
!cat > debug_inference.py << 'EOF'
import torch
from RFT_LM import RFTLM, MemoryBank
from transformers import AutoTokenizer

device = torch.device('cuda')

# Load fine-tuned RFT model
ck = torch.load('/zeng_gk/Amine/Huawei Challenge/RFT/AmineHL/rft_finetuned_retrieval_stable/rft_lm/best_model.pt', map_location='cpu')
a = ck['args']
model = RFTLM(
    vocab_size=a['vocab_size'], d_model=a['d_model'], n_layers=a['n_layers'],
    n_heads=a['n_heads'], window_size=a['window_size'], ff_mult=a['ff_mult'],
    memory_layer_idx=a['memory_layer_idx'], use_memory=True,
    mem_top_m=a['mem_top_m'], ocr_dim=a['ocr_dim'],
).to(device)
sd = ck['model']
if any(k.startswith('module.') for k in sd):
    sd = {k.replace('module.', ''): v for k, v in sd.items()}
model.load_state_dict(sd, strict=False)
model.eval()

tokenizer = AutoTokenizer.from_pretrained('/zeng_gk/Amine/Huawei Challenge/RFT/AmineHL/gpt2_tokenizer')

# Create a simple test input (2 chunks to force memory usage)
test_text = "The magic number alpha12345 is 42. This is some filler text to make the context longer. The magic number alpha12345 is"
input_ids = tokenizer.encode(test_text, return_tensors='pt').to(device)
chunk_size = 512
total_len = input_ids.shape[1]

print('=== RFT Inference Debug ===')
print(f'Input length: {total_len} tokens')
print(f'Chunk size: {chunk_size}')

mb = MemoryBank(max_entries=65536)
mb.reset()

# Process input in chunks
last_logits = None
for s in range(0, total_len, chunk_size):
    e = min(total_len, s + chunk_size)
    chunk = input_ids[:, s:e]
    mem_k, mem_v, mem_pos = mb.get_state()
    
    print(f'\nProcessing chunk {s//chunk_size + 1}: tokens {s}->{e}')
    print(f'  Memory bank size before chunk: {mb.size}')
    
    out = model(
        chunk,
        memory_keys=mem_k,
        memory_vals=mem_v,
        memory_positions=mem_pos,
        pos_offset=s,
        return_memory_state=True,
    )
    
    if 'new_mem_keys' in out:
        mb.add(out['new_mem_keys'].detach(), out['new_mem_vals'].detach(), s, e-s)
        print(f'  Added {e-s} entries to memory bank')
    
    last_logits = out['logits']

print(f'\nFinal memory bank size: {mb.size}')
print('===========================')

# Generate one token to see what it predicts
logits = last_logits[:, -1, :]
nxt = torch.argmax(logits, dim=-1)
pred_token = tokenizer.decode(nxt.item())
print(f'\nNext token prediction: "{pred_token}" (id={nxt.item()})')
print(f'True next token should be: "42"')
EOF

# Run debug script
!python debug_inference.py
```
